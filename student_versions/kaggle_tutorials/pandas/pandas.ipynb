# Index-based selection

`.iloc` is the **integer-position-based** way of selecting rows and columns from a DataFrame

```python
df.iloc[0]
```

You may also pass on slicing and lists to an index selection

```python
df.iloc[-5:] # Give me the last 5 rows of the entire DataFrame
```

```python
df.iloc[:3, 0] # Give me rows 0 through 2 for column 0, same as df.iloc[0:3, 0] or df.iloc[[0, 1, 2], 0]
```

# Label-based selection

`.loc` is a **label-based** way of selecting rows and columns from a DataFrame

Since `.loc` selects per labels, all labels in the slice will be included (e.g. `loc[0:10]` → selects labels 0 through 10, and `loc["flowers":"plants"]` →  selects labels from flowers through plants (position still matters)

```python
df.loc[5]
```

# Conditional selection

You can perform boolean operations over the DataFrame for filters

```python
print(df.c.shape)
df.c == 3 # "For each row, is the value in column c equal to 3?
```

```python
print(df.loc[df.c == 3].shape)
df.loc[df.c == 3] # Select/filter the rows where column c equals 3
```

You can also concatenate selections together as arguments for `.loc`

```python
df.loc[(df.c == 3) & (df.b >= 1)] # Select/filter rows where column c equals 3 AND column b is greater than or equal to 1
```

`isin` lets you select data whose value "is in" a list of values

```python
df.loc[df.c.isin([1, 3])] # Select/filter rows where column c contains 1 OR 3
```

`isnull` highlight values which are (or are not) empty (`NaN`)

```python
df.loc[df.c.notnull()] # Select rows where elements are not null (NaN)
```

# Maps

`map()` can go through each value and transform it based on the given mapping or function

```python
df_b_mean = df.b.mean()
df.b.map(lambda p:p - df_b_mean) # Take p (element in column b) and return p - df_b_mean
```

`apply()` is the equivalent method if we want to transform a whole DataFrame by calling a custom method on each row

```python
def remean_b(row):
  row.b = row.b - df_b_mean
  return row

df.apply(remean_b, axis='columns') # Apply remean_b to each row (of column b specified within remean_b), subtracting df_b_mean from column b
                                   # axis='index' applies the function to each column instead of each row (can't do here since row.b specifically means columns)
```

`map()` and `apply()` return transformed copies; they do not modify the original DataFrame in place.

pandas allows arithmetic between Series and DataFrames by aligning matching index/column labels

```python
df_b_mean = df.b.mean()
df.b - df_b_mean # Subtract df_b_mean from each value in column b
```

```python
df.c + '-' + df.message # Concatenate column c and column message via hypen
```

# Groupwise analysis

`groupby()` groups rows that have the same value(s) in the specified column(s)

`value_counts()` is just a shortcut to this `groupby()` operation

```python
df.groupby('b').b.count() # Equivalent of df.b.value_counts()
```

```python
df.groupby('message').b.min() # Group the rows by each unique value in message, then find the minimum value of b within each group
```

```python
df.groupby('message').apply(lambda df: df.c.iloc[0]) # For each message group, take the first value in column a
```

you can also group by more than one column with a simple list argument

```python
df.groupby(['c', 'message']).apply(lambda df: df.loc[df.b.idxmax()]) # For each unique combination of c and message, return the entire row where column b has its highest value
```

`agg()` can run many different functions on a DataFrame simultaneously

```python
df.groupby(['message']).c.agg([len, min, max]) # Returns the length, maximum value, and minimum value of elements in column c for each unique element in column message
```

# Multi-indexes

A multi-index differs from a regular index in that it has multiple levels, akin to a PivotTable

* Multi-indices have several methods for dealing with their tiered structure which are absent for single-level indices
* They also require two levels of labels to retrieve a value
* Dealing with multi-index output is a common "gotcha" for users new to pandas

```python
messages_df = df.groupby(['message', 'c']).b.agg([len]) # For each unique combination of column message and c, returns the number of values in column b
print(type(messages_df.index))
messages_df
```

The most popular multi-index method is for converting back to a regular index, the `reset_index()` method

```python
messages_df.reset_index()
```

# Sorting

`sort_values()` defaults to an ascending sort, where the lowest values go first

```python
messages_df = messages_df.reset_index()
messages_df.sort_values(by='len') # Sort by column len
```

However, most of the time we want a descending sort, where the higher numbers go first

```python
messages_df.sort_index(ascending=False) # Sort index in descending order
```

Better than Excel, you can sort by more than one column at a time

```python
messages_df.sort_values(by=['message', 'c']) # Sort by column message, then by column c
```

# Renaming index / columns

You can use `rename()` to change index names and/or column names via keyword arguments

```python
df.rename(columns={'message': 'msg'}) # Rename column message to msg
```

Here is an example of index rename via positional calls

```python
df = df.reset_index(drop=True) # drop=True tells pandas not to keep the old index when resetting it, so we effectively got rid of a in here (else we will not have index=0)
df.rename(index={0: 'firstEntry', 1: 'secondEntry'}) # Rename index a to firstEntry, 1 to secondEntry
```

Both the row index and the column index can have their own name attribute

The complimentary `rename_axis()` method may be used to change these names

```python
df.rename_axis("beta", axis='rows')
```
