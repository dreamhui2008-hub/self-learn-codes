# 1. Mechanical Refresher

*   In Python, a function is an object:
> *   you can store it in a variable,
> *   pass it into another function,
> *   return it from a function,
> *   and call it later with `()`.

*   A closure is an inner function that keeps access to variables from the outer function call that created it
> *   `*args` / `**kwargs` collect or unpack positional and keyword arguments when you need a forwarding function.
> *   `*args` stores variable positional arguments (as a tuple).
> *   `*kwargs` stores variable keyword arguments (as a dictionary).

# 2. Minimal Working Example

*   `apply` receives a function object and calls it later.
*   `make_offset` returns the inner `offset` function instead of calling it immediately.
*   `offset` still remembers `delta` after `make_offset` has finished.

```python
def twice(x):
  return x * 2

def apply(func, value):
  return func(value) # Calls whatever function that was passed in. So apply(twice, 5) is twice(5)

def make_offset(delta):
  def offset(x):
    return x + delta
  return offset

add_three = make_offset(3) # Remembers delta = 3
print(apply(twice, 5))
print(add_three(10)) # Calls offset(10), or x = 10 and x + delta, which was 3, and 10 + 3 = 13
```

```python
def increment(x):
   return x + 1

def apply_to_all(func, values):
  result = []
  for value in values:
    result.append(increment(value))
  return result

actual = apply_to_all(increment, [1, 2, 3])
expected = [2, 3, 4]
print("expected:", expected, "actual:", actual, "match:", actual == expected)
```

**Exercise 2.** Implement `make_clamper(low, high)` as a closure. Expected behavior: values stay inside `[0, 10]`.

```python
def make_clamper(low, high):
  def clamp(value):
    if value < low:
      return low
    if value > high:
      return high
    return value
  return clamp

clamp_score = make_clamper(0, 10)
actual = [clamp_score(-3), clamp_score(7), clamp_score(15)]
expected = [0, 7, 10]
print("expected:", expected, "actual:", actual, "match:", actual == expected)
```

**Exercise 3.** Implement `forward_call` so it forwards all positional and keyword arguments. Expected behavior: `12`.

```python
def weighted_sum(a, b, weight=1):
  return (a + b) * weight

def forward_call(func, *args, **kwargs):
  return func(*args, **kwargs)

actual = forward_call(weighted_sum, 2, 4, weight=2)
expected = 12
print("expected:", expected, "actual:", actual, "match:", actual == expected)
```

**Exercise 4.** Use argument unpacking to call `make_label`. Expected behavior: `'run-7-train'`.

# 7. Applied AI/ML Drill

**ML to Python mirror:**

choosing a metric function for predictions is just passing one function object into another function that will call it later.

**Python to ML mirror:**

real training and evaluation loops often accept loss functions, metric functions, or callbacks this same way;

the ML name changes, but the Python mechanic is still function objects plus argument passing.

**Applied Drill.** Implement `evaluate_metric` so it calls the metric function it receives. Expected behavior: MAE is `0.8333333333333334`.

```python
def mean_absolute_error(predictions, targets):
  total = 0
  for pred, target in zip(predictions, targets):
    total = total + abs(pred - target)
  return total / len(predictions)

def evaluate_metric(metric_fn, predictions, targets):
  return metric_fn(predictions, targets)

preds = [2.0, 3.0, 4.0]
targets = [1.0, 3.5, 5.0]
actual = evaluate_metric(mean_absolute_error, preds, targets)
expected = 0.8333333333333334
print("expected:", expected, "actual:", actual, "match:", actual == expected)
```

# 8. Common Bugs

- Calling too early: `func = square(4)` stores the result, not the function; the symptom is usually `TypeError: 'int' object is not callable`.
- Forgetting `return` for an inner function: the outer factory returns `None`; the symptom appears later when you try to call the returned value.
- Forgetting `*` while forwarding arguments: the callee receives one tuple instead of several arguments; the symptom is a wrong argument count or wrong output.
- Mutating captured data without meaning to: a closure can keep a list or dictionary alive; the symptom is state from a previous call appearing in a later call.

# 9. Compounding Drill

Combine closures with `*args` and `**kwargs`:

build a metric logger that remembers every result while forwarding arguments to the metric function.

```python
def make_metric_logger(metric_fn):
  history = []
  def logged(*args, **kwargs):
    history.append(difference(*args))
    return metric_fn(*args, **kwargs), history
  return logged

def difference(a, b):
  return abs(a - b)

logged_difference = make_metric_logger(difference)
first, history = logged_difference(10, 7)
second, history = logged_difference(3, 8)
expected = (3, 5, [3, 5])
actual = (first, second, history)
print("expected:", expected, "actual:", actual, "match:", actual == expected)
```
