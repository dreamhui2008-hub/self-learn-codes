## 1. Mechanical Refresher

* A Python instance stores its own attributes, and a method call like `obj.method(x)` is mechanically a function call where Python passes `obj` as the first argument named `self`.
* Class attributes live on the class and are found by instances when the instance does not have that attribute; `classmethod`, `staticmethod`, and `property` are built-in wrappers that change how attribute lookup turns class functions into callable or computed attributes.

### Python Classes: When to Use `self`, Class Attributes, `@classmethod`, `@staticmethod`, and `@property`

There are **two independent decisions** you make when designing a class:

1. **Where does the data live?**
2. **What is the method acting on?**

Keeping these separate makes Python's object model much easier to understand.

---

# 1. Where the data lives

## Instance attribute (`self.attribute`)

If **each object should have its own value**, use an instance attribute.

```python
class Player:
    def __init__(self):
        self.score = 0
```

Every player has their own score.

```python
alice = Player()
bob = Player()

alice.score += 10

print(alice.score)  # 10
print(bob.score)    # 0
```

**Use instance attributes when the data belongs to one object.**

Examples:

- Player score
- User name
- Bank account balance
- Car mileage

---

## Class attribute (`Class.attribute`)

If **every object should share one value**, use a class attribute.

```python
class Player:
    game_name = "Chess"
```

Every player sees the same `game_name`.

```python
print(Player.game_name)   # Chess

alice = Player()
bob = Player()

print(alice.game_name)    # Chess
print(bob.game_name)      # Chess
```

**Use class attributes when the data belongs to the class itself.**

Examples:

- Default settings
- Constants
- Population count
- Shared configuration

---

# 2. What the method acts on

## Instance method (`self`)

Use a normal method when you're working with **one specific object**.

```python
class Player:
    def __init__(self):
        self.score = 0

    def add_point(self):
        self.score += 1
```

```python
alice = Player()
alice.add_point()

print(alice.score)   # 1
```

Here you're changing **this player's** score.

### Rule of thumb

Use `self` when the method needs information from **one instance**.

Examples:

- Deposit into one bank account
- Move one game character
- Change one user's password

---

## Class method (`@classmethod`)

Use a class method when the operation is about the **class itself**, not one particular object.

```python
class Player:
    population = 0

    @classmethod
    def reset_population(cls):
        cls.population = 0
```

This method isn't about one player—it's about the `Player` class.

Another common example is a factory method:

```python
class Player:
    def __init__(self, name):
        self.name = name

    @classmethod
    def anonymous(cls):
        return cls("Anonymous")
```

```python
p = Player.anonymous()
```

Notice there is **no existing object** yet, so there is no `self`.

### Rule of thumb

Use `@classmethod` when the method needs the **class**, not an instance.

Examples:

- Factory methods
- Resetting shared state
- Tracking how many objects exist
- Alternate constructors

---

## Static method (`@staticmethod`)

Use a static method when the function **belongs conceptually to the class** but **doesn't need either `self` or `cls`.**

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b
```

```python
Math.add(2, 3)
```

or

```python
m = Math()
m.add(2, 3)
```

Both work exactly the same.

No object is modified.
No class data is accessed.

It's simply a function grouped inside the class.

### Rule of thumb

Use `@staticmethod` for helper functions that are closely related to the class but don't need its state.

Examples:

- Validation functions
- Parsing helpers
- Geometry calculations
- Unit conversions

---

## Property (`@property`)

Use a property when you want **a method to behave like an attribute**.

Without a property:

```python
class Circle:
    def area(self):
        return 3.14 * self.radius ** 2
```

You would write:

```python
c.area()
```

With a property:

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14 * self.radius ** 2
```

Now you write:

```python
c.area
```

even though Python is actually calling a method behind the scenes.

### Rule of thumb

Use `@property` when something should **feel like data** even though it is computed.

Examples:

- `person.full_name`
- `circle.area`
- `rectangle.perimeter`
- `employee.age`

---

# Quick Decision Guide

| If you need... | Use |
|---------------|-----|
| Data unique to each object | `self.attribute` |
| Data shared by every object | Class attribute |
| A method that works on one object | Instance method (`self`) |
| A method that works on the class | `@classmethod` |
| A helper function that doesn't need `self` or `cls` | `@staticmethod` |
| A computed value that should look like an attribute | `@property` |

# Mental Model

Think of it like houses.

- `self` = **one specific house**
- `cls` = **the blueprint for all houses**

Instance attributes (`self.x`) are things that belong to one house:

- Wall color
- Owner
- Furniture

Class attributes are things shared by every house built from the blueprint:

- Builder name
- Standard floor plan
- Building code version

Then the methods line up naturally:

- **Instance method (`self`)** → Do something to one house.
- **Class method (`cls`)** → Do something to the blueprint.
- **Static method** → A utility that's related to houses but doesn't need either a house or the blueprint.
- **Property** → A value that looks like a normal attribute but is calculated when you access it.

## 2. Minimal Working Example

*   `Counter.advance(counter)` shows the explicit form of what `counter.advance()` does for `self`.
*   `step` is found on the class unless an instance creates its own `step`.
> *   In this example, `step` is always 1.
*   `value` is stored directly on each instance (an instance method).


````python
class Counter:
  step = 1

  def __init__(self, start):
    self.value = start

  def advance(self):
    self.value = self.value + self.step
    return self.value

counter = Counter(10)
print(Counter.advance(counter)) # 10+1 = 11
print(counter.advance()) # Equivalent to Counter.advance(counter). 'counter' is automatically passed as self. value was 11, so 11 + 1 = 12.
print(counter.value) # 12
````

### Python Method Call Mental Models

## 1. Expanded Version (what Python is doing underneath)

When you write:

```python
Counter.advance(counter)
```

Think:

```
Class       Function       Instance
  ↓            ↓              ↓
Counter  .  advance  (  counter  )
```

Meaning:

- Find the `advance` function on the `Counter` class.
- Pass `counter` as the first argument.
- Inside the function, that argument is called `self`.

Equivalent internal view:

```python
def advance(self):
    ...
```

becomes:

```python
def advance(counter):
    ...
```

where `self` refers to `counter`.

---

## 2. Short Version (what Python programmers normally write)

When you write:

```python
counter.advance()
```

Think:

```
Instance       Method
    ↓             ↓
counter  .  advance  ( )
```

Python automatically expands it:

```
Instance       Class       Function       Instance
    ↓            ↓            ↓              ↓
counter  →  Counter  .  advance  (  counter  )
```

Meaning:

1. Start with the object (`counter`).
2. Look up its class (`Counter`).
3. Find the method (`advance`).
4. Automatically pass `counter` as `self`.

---

# Same Operation, Different Views

These are equivalent:

```python
counter.advance()
```

and

```python
Counter.advance(counter)
```

The difference is:

| Style | Shows | Best for |
|---|---|---|
| `counter.advance()` | Object performing an action | Writing normal Python code |
| `Counter.advance(counter)` | Class, function, and object relationship | Understanding `self` and debugging |

**Break-and-Fix Drill 1.** Break it by removing `self` from `rename`. Predict the argument-count error, then restore `self`.

````python
class Record:

  def __init__(self, name):
    self.name = name

  def rename(self, new_name): # Removing self from rename() -> TypeError: Record.rename() takes 1 positional argument but 2 were given
    self.name = new_name
    return self.name

record = Record("old")
actual = record.rename("new") # The record is automatically passed as self, and 'new' is passed as new_name. The method uses self to know which object's name to change (e.g. run rename() with "new")
expected = "new"
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````

**Break-and-Fix Drill 2.** Break it by assigning `Bucket.items = []` inside `add`.


Predict why separate instances share state, then fix it by using `self.items`.

````python
class Bucket:
  # To share, can simply add items = [] here without __init__()
  # If the instance has the attribute, Python uses it. If not, Python searches the class.

  def __init__(self):
    # Bucket.items = []
    self.items = []

  def add(self, item):
    # Bucket.items = []
    # Bucket.items.append(item)
    self.items.append(item)

first = Bucket()
second = Bucket()
first.add("a") # Equals to Bucket().add.("a")
actual = [first.items, second.items] # Still [["a"], []] because Bucket.items is never called. To share, use Bucket.items.append(item) in add() and use Bucket.items in __init__()
expected = [["a"], []]
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````

## 6. Standalone Exercises

**Exercise 1.** Fill in `full_name` using `self.first` and `self.last`. Expected behavior: `'Ada Lovelace'`.

````python
class Person:

  def __init__(self, first, last):
    self.first = first
    self.last = last

  def full_name(self):
    return self.first + " " + self.last

actual = Person("Ada", "Lovelace").full_name()
expected = "Ada Lovelace"
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````

**Exercise 2.** Add only to the instance attribute. Expected behavior: `[['x'], []]`.

````python
class Log:
  default_name = "run"

  def __init__(self):
    self.entries = []

  def add(self, entry):
    self.entries.append(entry)

a = Log()
b = Log()
a.add("x")
actual = [a.entries, b.entries]
expected = [["x"], []]
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````

**Exercise 3.** Create a property manually. Expected behavior: reading `area` returns `12`.

````python
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
    area = property(area) # Cannot be called as method without property()

actual = Rectangle(3, 4).area
expected = 12
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````

**Exercise 4.** Create a `classmethod` named `from_pair`. Expected behavior: the object stores `left=2` and `right=5`.

````python
class Pair:

    def __init__(self, left, right):
        self.left = left
        self.right = right

    def from_pair(cls, pair):
        return cls(pair[0], pair[1])
    from_pair = classmethod(from_pair)

actual_obj = Pair.from_pair((2, 5))
actual = [actual_obj.left, actual_obj.right]
expected = [2, 5]
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````

**Exercise 5.** Create a `staticmethod` named `valid_size`. Expected behavior: `[False, True]`.

````python
class Batch:

    def valid_size(size):
        return size > 0
    valid_size = staticmethod(valid_size)

actual = [Batch().valid_size(0), Batch().valid_size(3)]
expected = [False, True]
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````

**Applied Drill.** Complete the computed `parameter_count` property without decorator syntax `@property`.

Expected behavior: a linear layer with 3 inputs and 2 outputs has 8 parameters.

````python
class LinearConfig:

    def __init__(self, inputs, outputs):
        self.inputs = inputs
        self.outputs = outputs

    def parameter_count(self):
        return self.outputs ** self.inputs

    parameter_count = property(parameter_count)

config = LinearConfig(3, 2)
actual = config.parameter_count
expected = 8
print("expected:", expected, "actual:", actual, "match:", actual == expected)


# You need to calculate 8 params because (3 weights for each layer * 2 + 1 bias for each layer * 2, or 6+2, or 8)
# Inputs                 Outputs

# x1 ───────────────┐
# x2 ───────────────┼──> y1
# x3 ───────────────┘


# x1 ───────────────┐
# x2 ───────────────┼──> y2
# x3 ───────────────┘
````

## 8. Common Bugs

- **Missing `self`**: instance method calls pass the instance automatically; the symptom is an argument-count `TypeError`.
- **Writing to the class when you meant the instance**: the symptom is state leaking across objects.
- **Confusing `obj.method` with `obj.method()`**: the first is a bound method object; the second calls it and returns its result.
- **Forgetting that property access has no parentheses**: the symptom is either a method object printed or a `'int' object is not callable` error after adding extra parentheses.

### Python: Dot Access vs Parentheses

Normally, a method needs parentheses to run.

```python
obj.method
```

means:

> "Give me the method itself."

It does **not** execute the code.

```python
obj.method()
```

means:

> "Run the method and give me the result."

---

## The `@property` exception

A normal method needs parentheses to run:

```python
obj.method()
```

But if you wrap a method with `property`, the method becomes accessible like an attribute.

Decorator form:

```python
class Example:
    @property
    def value(self):
        return 10
```

The decorator is just a shorter way of writing:

```python
class Example:
    def value(self):
        return 10

    value = property(value)
```

The inner function is created first, then `property()` wraps it and replaces the original method name with a property object.

Now:

```python
example.value
```

runs the hidden function and returns:

```text
10
```

You no longer write:

```python
example.value()
```

because `example.value` is no longer returning the function — it is returning the function's result.

---

A useful mental model:

Without `property`:

```python
obj.method
```

→ "Give me the function"

```python
obj.method()
```

→ "Run the function"

With `property`:

```python
obj.value
```

→ "Run the wrapped function and give me the result"

## 9. Compounding Drill

Combine Chapter 1 function objects with this chapter's object model:

store a metric function on an evaluator instance and call it from a method.

````python
def absolute_error(pred, target):
    return abs(pred - target)

class Evaluator:

    def __init__(self, metric_fn):
        self.metric_fn = metric_fn

    def evaluate_one(self, pred, target):
        return absolute_error(pred, target)

evaluator = Evaluator(absolute_error)
actual = evaluator.evaluate_one(10, 6)
expected = 4
print("expected:", expected, "actual:", actual, "match:", actual == expected)
````
